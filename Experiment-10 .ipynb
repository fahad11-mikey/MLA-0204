```python
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs

X, _ = make_blobs(n_samples=300, centers=3, random_state=42, cluster_std=1.2)

model = GaussianMixture(n_components=3, random_state=42)
model.fit(X)
labels = model.predict(X)

print("Means of Gaussian components:\n", model.means_)
print("Weights:", model.weights_)
print("Converged:", model.converged_)
print("Number of EM iterations:", model.n_iter_)

plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.scatter(model.means_[:, 0], model.means_[:, 1],
            c='red', marker='X', s=200, label='Component means')
plt.title("Expectation Maximization Clustering")
plt.legend()
plt.show()
```
